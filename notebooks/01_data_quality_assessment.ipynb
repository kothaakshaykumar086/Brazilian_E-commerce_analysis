```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw")

csv_files = sorted(DATA_DIR.glob("*.csv"))

print(f"Number of CSV files found: {len(csv_files)}")

for file in csv_files:
    print(file.name )
```

```python
for file in csv_files:
    df = pd.read_csv(file)

    print(f"\n{'=' * 60}") ## prints 60 = signs to separate the output for each file
    print(f"File: {file.name}")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
    print("\nColumn names:")
    print(df.columns.tolist()) ## converts the column names to a list and prints them
```

```python
# create loop, read each CSV file,
#  print "=", print file name,print missing values,
#  print duplicates

for file in csv_files:
    df = pd.read_csv(file)

    print(f"\n{'='*60}")
    print(f"File_Name: {file.name}")

    # check missing values
    print("\nMissing Values:")
    missing = df.isnull().sum()
    print(missing[missing > 0])

    # check duplicated rows
    print("\nDuplicate Rows:",df.duplicated().sum())
```

```python
for file in csv_files:
    df = pd.read_csv(file)
    print(f"\n{'='*60}")
    print(f"File_Name: {file.name}")
    print(df.dtypes)
```

```python

for file in csv_files:
    df = pd.read_csv(file)

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if not missing.empty:
        print(f"\n{'=' * 60}")
        print(f"File: {file.name}")
        print("Missing-value count by column:")
        print(missing.sort_values(ascending=False))

        print("\nMissing-value percentage:")
        print((missing / len(df) * 100).round(2))
```

```python

orders_file = next(
    file for file in csv_files
    if file.name == "olist_orders_dataset.csv"
)

orders = pd.read_csv(orders_file)

print("Order counts by status:")
print(orders["order_status"].value_counts(dropna=False))

print("\nMissing timestamps by order status:")
timestamp_cols = [
    col for col in orders.columns
    if "date" in col or "time" in col
]

print(
    orders.groupby("order_status")[timestamp_cols]
    .apply(lambda group: group.isnull().sum())
)
```

```python

# Select orders that have missing delivery timestamps
missing_delivery = orders[
    orders["order_delivered_customer_date"].isnull()
]

# Count these orders by their status
print("Orders missing customer delivery date:")
print(missing_delivery["order_status"].value_counts())

# Check whether delivered orders are missing delivery timestamps
delivered_missing = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].isnull())
]

print(
    "\nDelivered orders missing delivery date:",
    len(delivered_missing)
)
```

```python

# Select delivered orders with a missing customer delivery date
delivered_missing = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].isnull())
]

# Display the order IDs and other timestamps to investigate the issue
print(
    delivered_missing[
        [
            "order_id",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ].to_string(index=False)
)
```

```python
# Identify columns that contain date or time information
date_cols = [
    col for col in orders.columns
    if "date" in col or "time" in col
]

# Display their current data types
print(orders[date_cols].dtypes)
```

```python

# Create a copy so the original DataFrame stays unchanged
orders_clean = orders.copy()

# Identify the timestamp columns we want to convert
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

# Convert each selected column from text to datetime
for col in date_cols:
    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors="coerce"
    )

# Display the updated data types
print(orders_clean[date_cols].dtypes)
```

```python
# Check the number of missing values after conversion
print(orders_clean[date_cols].isnull().sum())

# Check that the original DataFrame still has its original types
print("\nOriginal purchase timestamp type:")
print(orders["order_purchase_timestamp"].dtype)
```

```python
# Compare missing values before and after datetime conversion
print("Original missing-value counts:")
print(orders[date_cols].isnull().sum())

print("\nCleaned missing-value counts:")
print(orders_clean[date_cols].isnull().sum())

# Verify the original DataFrame's purchase timestamp type
print("\nOriginal purchase timestamp type:")
print(orders["order_purchase_timestamp"].dtype)

# Verify the cleaned purchase timestamp type
print("\nCleaned purchase timestamp type:")
print(orders_clean["order_purchase_timestamp"].dtype)
```

```python

# Locate the products CSV file
products_file = next(
    file for file in csv_files
    if file.name == "olist_products_dataset.csv"
)

# Load the original products data
products = pd.read_csv(products_file)

# Create a separate copy for future cleaning
products_clean = products.copy()

# Count missing values in every column
product_missing = products_clean.isnull().sum()

# Keep only columns that contain missing values
product_missing = product_missing[product_missing > 0]

# Sort from the most missing values to the fewest
print("Missing values by product column:")
print(product_missing.sort_values(ascending=False))

# Calculate the percentage of missing values in each affected column
print("\nMissing-value percentage:")
print(
    (product_missing / len(products_clean) * 100)
    .round(2)
    .sort_values(ascending=False)
)
```

```python

# Columns with 610 missing values
main_missing_cols = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
]

# Check whether all four columns are missing in the same rows
all_four_missing = products_clean[main_missing_cols].isnull().all(axis=1)

print("Products missing all four attributes:", all_four_missing.sum())

# Check whether any rows have only some of these four values missing
some_missing = (
    products_clean[main_missing_cols].isnull().any(axis=1)
    & ~products_clean[main_missing_cols].isnull().all(axis=1)
)

print("Products missing only some of the four attributes:", some_missing.sum())

# Check the two products with missing physical measurements
dimension_cols = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

missing_dimensions = products_clean[
    products_clean[dimension_cols].isnull().any(axis=1)
]

print("\nProducts with missing physical measurements:")
print(missing_dimensions[["product_id"] + dimension_cols].to_string(index=False))
```

```python

# Locate the geolocation CSV file
geo_file = next(
    file for file in csv_files
    if file.name == "olist_geolocation_dataset.csv"
)

# Load the original geolocation data
geolocation = pd.read_csv(geo_file)

# Count completely duplicated rows
print(
    "Completely duplicated rows:",
    geolocation.duplicated().sum()
)

# Count unique values in each column
print("\nUnique values per column:")
print(geolocation.nunique())

# Check how many unique postal-code prefixes exist
print(
    "\nUnique ZIP-code prefixes:",
    geolocation["geolocation_zip_code_prefix"].nunique()
)

# Check how many unique latitude-longitude pairs exist
print(
    "Unique coordinate pairs:",
    geolocation[
        ["geolocation_lat", "geolocation_lng"]
    ].drop_duplicates().shape[0]
)
```

```python
# Create a cleaned copy without completely identical rows
geolocation_clean = geolocation.drop_duplicates().copy()

# Compare the original and cleaned row counts
print("Original row count:", len(geolocation))
print("Cleaned row count:", len(geolocation_clean))

# Confirm that exact duplicates have been removed
print(
    "Remaining exact duplicates:",
    geolocation_clean.duplicated().sum()
)

# Calculate how many rows were removed
removed_rows = len(geolocation) - len(geolocation_clean)

print("Duplicate rows removed:", removed_rows)
print(
    "Percentage of rows removed:",
    round(removed_rows / len(geolocation) * 100, 2),
    "%"
)
```

```python
# Count how many different coordinate pairs belong to each ZIP prefix
coordinates_per_zip = (
    geolocation_clean
    .groupby("geolocation_zip_code_prefix")[
        ["geolocation_lat", "geolocation_lng"]
    ]
    .apply(lambda group: group.drop_duplicates().shape[0])
)

# Summarize the number of coordinate pairs per ZIP prefix
print("ZIP prefixes with multiple coordinate pairs:",
      (coordinates_per_zip > 1).sum())

print("\nCoordinate pairs per ZIP prefix:")
print(coordinates_per_zip.describe())
```

```python
# Locate the customer reviews CSV file
reviews_file = next(
    file for file in csv_files
    if file.name == "olist_order_reviews_dataset.csv"
)

# Load the reviews dataset
reviews = pd.read_csv(reviews_file)

# Check dataset dimensions
print("Dataset shape:", reviews.shape)

# Check missing values in each column
print("\nMissing values per column:")
print(reviews.isnull().sum().sort_values(ascending=False))

# Check exact duplicate rows
print("\nCompletely duplicated rows:", reviews.duplicated().sum())

# Check customer review score distribution
print("\nReview score distribution:")
print(reviews["review_score"].value_counts().sort_index())

# Check whether an order has multiple review records
print(
    "\nOrders with multiple review records:",
    (reviews["order_id"].value_counts() > 1).sum()
)
```

```python
# Locate and load the order payments dataset
payments_file = next(
    file for file in csv_files
    if file.name == "olist_order_payments_dataset.csv"
)

payments = pd.read_csv(payments_file)

# Check dataset dimensions
print("Dataset shape:", payments.shape)

# Check missing values
print("\nMissing values:")
print(payments.isnull().sum())

# Check exact duplicate rows
print("\nCompletely duplicated rows:", payments.duplicated().sum())

# Examine payment methods
print("\nPayment type counts:")
print(payments["payment_type"].value_counts(dropna=False))

# Summarize payment values
print("\nPayment value summary:")
print(payments["payment_value"].describe())

# Check installment counts
print("\nInstallment count distribution:")
print(payments["payment_installments"].value_counts().sort_index())
```

```python
# Locate and load the order items dataset
items_file = next(
    file for file in csv_files
    if file.name == "olist_order_items_dataset.csv"
)

items = pd.read_csv(items_file)

# Check dataset dimensions
print("Dataset shape:", items.shape)

# Check missing values
print("\nMissing values:")
print(items.isnull().sum())

# Check exact duplicate rows
print("\nCompletely duplicated rows:", items.duplicated().sum())

# Summarize item prices
print("\nPrice summary:")
print(items["price"].describe())

# Summarize freight values
print("\nFreight value summary:")
print(items["freight_value"].describe())

# Check how many orders contain multiple item records
items_per_order = items["order_id"].value_counts()

print(
    "\nOrders with multiple item records:",
    (items_per_order > 1).sum()
)

# Check unusual zero or negative values
print("\nZero or negative prices:", (items["price"] <= 0).sum())
print(
    "Negative freight values:",
    (items["freight_value"] < 0).sum()
)
```
